## Vectors in NumPy

---

In Lab 1, you looked at the popularity of a name over time: one number per year, in order. As we saw in Lecture 2 and Chapter 1.4, an ordered list of numbers is called a **vector**.

Python has a built-in data structure called a `list`, created using square brackets. Let's see what happens when we add two lists.


````python
[1, 2, 3] + [4, 5, 6]
````

Python joined the two lists instead of adding their entries. This is not vector addition!

**NumPy** is a Python module that allows us to work with vectors. As in lab, we need to import it before using it. We'll use the nickname `np`.


````python
import numpy as np
````

Instead of importing just a single function, as we did in the Lab 1 notebook, we imported an entire module. To use a function from this module, we put `np.` before its name.

The first function we'll use is `np.array`. It takes a list of numbers and returns an array. Run the cell below.


````python
v = np.array([5, 9, -3, 26])
v
````

NumPy calls the object above an **array**. For now, we'll use arrays to represent vectors.

`len(v)` gives the number of entries in `v`. To access a particular entry, we use square brackets. Python starts counting at **zero**, so `v[0]` gives the first entry, `v[1]` gives the second, and so on.


````python
# This is not the same as the length of the vector!
len(v)
````

````python
# First entry
v[0]
````

````python
# Third entry
v[2]
````

Observe what happens below:

````python
v = np.array([1, 2, 3])
w = np.array([4, 5, 6])

# Element-wise addition
v + w
````

This time, `v + w` produced an array with 3 entries, each equal to the sum of the corresponding entries in `v` and `w`. This is vector addition!

````python
# Scalar multiplication
3 * v
````

````python
v / 2
````

````python
v + 10
````

Notice that `v + 10` added `10` to **each entry** of `v`. Here, Python allows us to add a number to an array, even though we haven't defined addition between a scalar and a vector in class.

How can we represent a **text document as a vector**? In text analysis, each piece of text we want to analyze is called a **document**, and a collection of documents is called a **corpus**. ### The Bag of Words Model

One simple idea is to count the number of times each word appears in a document. Consider these three documents:

1. **math math math math engineering class**
2. **engineering math engineering michigan**
3. **michigan math engineering**

Across these documents, there are four different words. We'll always use the order **math, engineering, class, michigan**, so each document becomes a vector in $\mathbb{R}^4$:

$$\vec v=\begin{bmatrix}
\text{count of "math"}\\
\text{count of "engineering"}\\
\text{count of "class"}\\
\text{count of "michigan"}
\end{bmatrix}.$$

This is called the **bag of words** model because it ignores the order of words within a document. Imagine putting the words in a bag and shaking it: their counts stay the same. However, the order of the **coordinates** must be the same for every vector, so that corresponding entries count the same word.

## Cosine Similarity

---

Recall that the cosine similarity of two nonzero vectors is

$$\cos\theta = \frac{\vec u\cdot\vec v}{\|\vec u\|\|\vec v\|}.$$

Here, $\theta$ is the angle between the two vectors. Cosine similarity measures how closely their **directions** align. A value of 1 means the vectors point in the same direction, and a value of 0 means they are orthogonal. Because word counts are nonnegative, the cosine similarities in this activity will be between 0 and 1.

## Comparing Presidential Speeches

---

Each year, the sitting US President gives a speech, typically called the State of the Union address. Now, we'll apply the bag of words model and cosine similarity to a collection of State of the Union messages and related addresses to Congress from 1790 through 2025. The transcripts come from [The American Presidency Project](https://www.presidency.ucsb.edu/documents/presidential-documents-archive-guidebook/annual-messages-congress-the-state-the-union).

We'll use the 500 most common words across the collection, always in the same coordinate order. Each speech will be represented by a vector in $\mathbb{R}^{500}$, with one count per word. We treat uppercase and lowercase versions of a word as the same word and ignore punctuation.

**Run the provided setup cell below. You do not need to write, change, or understand its code.** It loads all 235 speeches and prepares their word-count vectors.


````python
#@title Load the speeches and prepare their vectors { display-mode: "form" }
from pathlib import Path
from urllib.request import urlopen

data_path = Path('data/stateoftheunion1790-2025.txt')
if data_path.exists():
    sotu = data_path.read_text(encoding='utf-8')
else:
    data_url = ('https://raw.githubusercontent.com/math-124/fa26-code/main/'
                'homeworks/hw02/data/stateoftheunion1790-2025.txt')
    with urlopen(data_url) as response:
        sotu = response.read().decode('utf-8')


from collections import Counter
from datetime import datetime
import re

speech_records = []
for speech in sotu.split('\n***\n')[1:]:
    parts = speech.strip().split('\n', maxsplit=3)
    text = re.sub(r"[^A-Za-z' ]", ' ', parts[3]).lower()
    text = re.sub(r"it's", 'it is', text).replace(' s ', '')
    speech_records.append({
        'name': parts[1].strip() + ': ' + parts[2].strip(),
        'president': parts[1].strip(),
        'date': datetime.strptime(parts[2].strip(), '%B %d, %Y'),
        'transcript': speech.strip(),
        'counts': Counter(text.split()),
    })
speech_records.sort(key=lambda record: record['date'])
speech_names = [record['name'] for record in speech_records]
speech_dates = [record['date'].strftime('%Y-%m-%d') for record in speech_records]
speech_labels = [date + ' — ' + record['president']
                 for date, record in zip(speech_dates, speech_records)]

total_counts = Counter()
for record in speech_records:
    total_counts.update(record['counts'])
terms = [word for word, count in total_counts.most_common(500)]
speech_vectors = np.array([[record['counts'][word] for word in terms]
                           for record in speech_records])
print(f'Loaded {len(speech_records)} speeches, each represented by {len(terms)} word counts.')
````

### Explore Similarity over Time

Run the next cell, then use the graph's **reference speech** menu to choose any speech in the collection. (You may need to scroll through the menu by clicking and dragging the scroll bar.) Each point compares that reference speech with another speech, positioned by date. Hover over a point to see its president, date, and cosine similarity with the reference speech.

The reference speech itself is omitted: its cosine similarity with itself would be 1. The dashed vertical line marks its date. Gold points are other speeches by the same president; blue points are speeches by other presidents. You can zoom in by dragging over part of the plot and double-click to reset.

````python
#@title Explore cosine similarity over time { display-mode: "form" }
import plotly.graph_objects as go

speech_similarities = np.array([
    [cosine_similarity(u, v) for v in speech_vectors]
    for u in speech_vectors
], dtype=float)
speech_similarities = np.clip(speech_similarities, -1, 1)
initial_reference = len(speech_records) - 1

def reference_values(reference):
    return [None if i == reference else float(value)
            for i, value in enumerate(speech_similarities[reference])]

def reference_colors(reference):
    president = speech_records[reference]['president']
    return ['#B8860B' if record['president'] == president else '#2467A0'
            for record in speech_records]

def reference_layout(reference):
    return dict(
        shapes=[dict(type='line', xref='x', yref='paper',
                     x0=speech_dates[reference], x1=speech_dates[reference],
                     y0=0, y1=1, line=dict(color='#666666', width=1.5, dash='dash'))],
        annotations=[
            dict(text='Reference speech:', x=0, y=1.28, xref='paper', yref='paper',
                 xanchor='left', showarrow=False),
            dict(text='Gold: same president · Blue: other presidents',
                 x=1, y=-0.22, xref='paper', yref='paper', xanchor='right', showarrow=False),
        ],
    )

speech_figure = go.Figure(go.Scatter(
    x=speech_dates, y=reference_values(initial_reference),
    customdata=speech_names, mode='lines+markers', connectgaps=True,
    line=dict(color='#C5CDD4', width=1),
    marker=dict(size=7, color=reference_colors(initial_reference)),
    hovertemplate='%{customdata}<br>Cosine similarity: %{y:.4f}<extra></extra>',
))
lower_limit = max(0, float(np.floor((speech_similarities.min()-0.02)*10)/10))
speech_figure.update_layout(
    title=dict(text='Speech Similarity over Time', x=0.06, y=0.98),
    template='plotly_white', font=dict(family='Palatino, Georgia, serif', size=14),
    height=620, margin=dict(l=70, r=35, t=150, b=110), showlegend=False,
    xaxis=dict(title='Date of the comparison speech', type='date'),
    yaxis=dict(title='Cosine similarity', range=[lower_limit, 1.005]),
    updatemenus=[dict(
        buttons=[dict(label=label, method='update',
                      args=[{'y':[reference_values(i)], 'marker.color':[reference_colors(i)]},
                            reference_layout(i)])
                 for i, label in enumerate(speech_labels)],
        active=initial_reference, direction='down', x=0, y=1.10,
        xanchor='left', yanchor='bottom', showactive=True,
    )],
    **reference_layout(initial_reference),
)
speech_figure.show(config={'displaylogo':False, 'responsive':True})
````

### Read the Speeches

Use the reader below to open any speech, including the reference speech and any interesting point on the graph. The reader preserves the original transcript's wording and punctuation.


````python
#@title Read a speech { display-mode: "form" }
import html
import ipywidgets as widgets
from IPython.display import display

# Enable interactive widgets when running in Google Colab.
try:
    from google.colab import output
    output.enable_custom_widget_manager()
except ImportError:
    pass

speech_reader_menu = widgets.Dropdown(
    options=[(label, i) for i, label in enumerate(speech_labels)],
    value=initial_reference, description='Read:',
    layout=widgets.Layout(width='95%'), style={'description_width':'initial'},
)
speech_reader_text = widgets.HTML()

def update_speech_reader(change=None):
    i = speech_reader_menu.value
    transcript = html.escape(speech_records[i]['transcript'])
    speech_reader_text.value = (
        '<div style="max-height:460px;overflow-y:auto;padding:18px;border:1px solid #ddd;'
        'font-family:Georgia,serif;line-height:1.6;background:white;color:#222;">'
        f'<div style="white-space:pre-wrap;">{transcript}</div></div>'
    )

speech_reader_menu.observe(update_speech_reader, names='value')
update_speech_reader()
display(widgets.VBox([speech_reader_menu, speech_reader_text]))
````
