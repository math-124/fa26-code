## From projection to reflection

---

In Homework 1, you changed the **colors** of pixels in an image. This time, you'll change their **positions** to create a mirror image, while keeping the colors the same.

**Goal: Implement the reflection formula from Problem 9, then use your function to reflect an image across a line that you control.**

We'll start with a single vector. Once your function works, the supplied code will let you upload an image and reflect it across different lines. The image lives in $\mathbb{R}^2$, but your function will also work for vectors in $\mathbb{R}^3$, just as in Problem 9.

Let $\vec v$ be the vector we want to reflect, and let $\vec w$ be a **nonzero direction vector** for a line through the origin. In Python, `w` is the name of this direction vector; the line itself is $\ell=\operatorname{span}(\vec w)$.

The projection of $\vec v$ onto the line is

$$\vec p=\frac{\vec v\cdot\vec w}{\vec w\cdot\vec w}\vec w.$$

The vector $\vec v-\vec p$ is the part perpendicular to the line. Reflection keeps $\vec p$ and reverses that perpendicular part, giving

$$\vec v_{\mathrm{ref}}=\vec p-(\vec v-\vec p)=2\vec p-\vec v.$$

This matches the picture in Problem 9: the tip of $\vec p$ is halfway between the tips of $\vec v$ and its reflection.

**The direction vector does not need to have length 1.** For example, $\begin{bmatrix}1\\1\end{bmatrix}$ and $\begin{bmatrix}2\\2\end{bmatrix}$ describe the same line, so they must give the same reflection. The denominator in the projection formula accounts for this.

### An image is a collection of positions

Think of each pixel's position as the tip of a vector $\begin{bmatrix}x\\y\end{bmatrix}$. Here, $x$ increases to the right and $y$ increases upward. Reflecting the image means reflecting these position vectors while keeping their colors.

The slider chooses an angle $\theta$, measured **counterclockwise from the positive $x$-axis**. The supplied code uses the direction vector

$$\vec w=\begin{bmatrix}\cos\theta\\\sin\theta\end{bmatrix}.$$

At $0^\circ$, the line is horizontal. At $90^\circ$, it is vertical. At $180^\circ$, it is the same horizontal line as at $0^\circ$, with the direction vector pointing the opposite way.

Run the next cell, then drag the slider. The **dotted blue line** is the line of reflection and passes through the origin. The image labeled **Original** stays upright. As you turn the line, the supplied code moves that image so it stays entirely on one side of the line, with its reflection on the other side. This movement is just how we arrange the picture; your function performs the reflection.

The supplied code uses your function to reflect three corners of the image, then draws the pixels between those reflected corners. This works because reflection preserves straight lines and evenly spaced positions.

Try $0^\circ$, $90^\circ$, and a tilted line such as $45^\circ$. At $0^\circ$, you should see a top-to-bottom flip; at $90^\circ$, a left-to-right flip. For a tilted line, look for matching features the same distance from opposite sides of the line.

````python
#@title Explore your image reflection { display-mode: "form" }
import base64
import json
import uuid

# Prepare the image and all slider positions using YOUR reflection function.
# The browser only draws the results, so dragging the slider needs no Python callback.
width, height = small_image.size
radius = np.hypot(width, height) / 2
padding = max(4, radius * 0.20)
frames = []
for theta_degrees in range(181):
    theta = np.deg2rad(theta_degrees)
    w = np.array([np.cos(theta), np.sin(theta)])
    normal = np.array([-w[1], w[0]])
    center = (radius + padding) * normal
    # Top-left, top-right, and bottom-left corners in x-right, y-up coordinates.
    corners = center + np.array([
        [-width / 2, height / 2],
        [width / 2, height / 2],
        [-width / 2, -height / 2],
    ])
    reflected = np.array([reflection(corner.copy(), w.copy()) for corner in corners], dtype=float)
    if reflected.shape != (3, 2) or not np.isfinite(reflected).all():
        raise ValueError('Check your reflection function and rerun Task 1 before continuing.')
    frames.append(dict(w=w.tolist(), original=corners.tolist(), reflected=reflected.tolist()))

image_buffer = io.BytesIO()
small_image.save(image_buffer, format='PNG')
image_base64 = base64.b64encode(image_buffer.getvalue()).decode('ascii')
canvas_template = r"""<div id="__ID__" style="max-width:1000px;margin:12px auto;color:#202020;font:17px Palatino,Georgia,serif">
  <div style="padding:18px;border:1px solid #d6dce2;border-radius:10px;background:#fff">
    <label style="display:block" for="__ID__-theta">Angle θ: <strong data-angle>45°</strong></label>
    <input id="__ID__-theta" aria-label="Line angle in degrees" type="range" min="0" max="180" step="1" value="45" style="width:100%;margin:12px 0;accent-color:#2467a0">
    <div style="display:flex;justify-content:space-between;font-size:14px;color:#555"><span>0° · horizontal</span><span>90° · vertical</span><span>180° · horizontal</span></div>
    <div style="display:flex;justify-content:center;align-items:center;gap:8px;margin-top:14px">
      <span>Direction vector</span>
      <math xmlns="http://www.w3.org/1998/Math/MathML" display="inline" style="font-size:21px">
        <mrow>
          <mover accent="true"><mi>w</mi><mo>→</mo></mover><mo>=</mo>
          <mrow><mo stretchy="true">[</mo>
            <mtable columnalign="center" rowspacing="0.25em">
              <mtr><mtd><mn data-wx>0.707</mn></mtd></mtr>
              <mtr><mtd><mn data-wy>0.707</mn></mtd></mtr>
            </mtable>
          <mo stretchy="true">]</mo></mrow>
        </mrow>
      </math>
    </div>
    <canvas width="1000" height="440" role="img" aria-label="Original image and its reflection on opposite sides of a dotted blue line through the origin" style="display:block;width:100%;height:auto;margin-top:8px"></canvas>
    <p style="margin:8px 0 0;font-size:15px;line-height:1.45;text-align:center">The original stays upright and moves as the line turns.<br>The reflected image is drawn using your <code>reflection</code> function.</p>
  </div>
</div>
<script>
(() => {
  const root = document.getElementById('__ID__');
  const frames = __FRAMES__;
  const extent = __EXTENT__;
  const w = __WIDTH__, h = __HEIGHT__;
  const canvas = root.querySelector('canvas');
  const ctx = canvas.getContext('2d');
  const slider = root.querySelector('input');
  const picture = new Image();
  const scale = Math.min(canvas.width - 100, canvas.height - 110) / (2 * extent);
  const screen = ([x, y]) => [canvas.width / 2 + scale*x, canvas.height / 2 - scale*y];
  function segment(a, b, color, width, dash=[]) {
    const A = screen(a), B = screen(b);
    ctx.beginPath(); ctx.setLineDash(dash); ctx.strokeStyle = color; ctx.lineWidth = width;
    ctx.moveTo(...A); ctx.lineTo(...B); ctx.stroke(); ctx.setLineDash([]);
  }
  function drawImage(corners, label) {
    const [A, B, C] = corners.map(screen);
    ctx.save();
    ctx.transform((B[0]-A[0])/w, (B[1]-A[1])/w,
                  (C[0]-A[0])/h, (C[1]-A[1])/h, A[0], A[1]);
    ctx.imageSmoothingEnabled = false;
    ctx.drawImage(picture, 0, 0, w, h);
    ctx.restore();
    const D = [B[0]+C[0]-A[0], B[1]+C[1]-A[1]];
    ctx.strokeStyle = '#777'; ctx.lineWidth = 1;
    ctx.beginPath(); ctx.moveTo(...A); ctx.lineTo(...B); ctx.lineTo(...D); ctx.lineTo(...C); ctx.closePath(); ctx.stroke();
    const centerX = (B[0]+C[0])/2;
    const bottom = Math.max(A[1],B[1],C[1],D[1]);
    ctx.font = '17px Palatino, Georgia, serif';
    ctx.textAlign = 'center'; ctx.textBaseline = 'top';
    const textWidth = ctx.measureText(label).width;
    ctx.fillStyle = 'rgba(255,255,255,0.94)'; ctx.fillRect(centerX-textWidth/2-5,bottom+5,textWidth+10,23);
    ctx.fillStyle = '#202020'; ctx.fillText(label, centerX, bottom+7);
  }
  function draw() {
    const theta = Number(slider.value), frame = frames[theta];
    root.querySelector('[data-angle]').textContent = theta + '°';
    const pretty = x => (Math.abs(x)<0.0005 ? 0 : x).toFixed(3);
    root.querySelector('[data-wx]').textContent = pretty(frame.w[0]);
    root.querySelector('[data-wy]').textContent = pretty(frame.w[1]);
    ctx.clearRect(0,0,canvas.width,canvas.height);
    ctx.fillStyle='#fff'; ctx.fillRect(0,0,canvas.width,canvas.height);
    segment([-extent,0],[extent,0],'#ddd',1);
    segment([0,-extent],[0,extent],'#ddd',1);
    ctx.font='15px Palatino, Georgia, serif'; ctx.fillStyle='#777'; ctx.textAlign='left'; ctx.textBaseline='middle';
    const X=screen([extent,0]), Y=screen([0,extent]);
    ctx.fillText('x',X[0]+9,X[1]); ctx.fillText('y',Y[0]+8,Y[1]);
    const end = frame.w.map(x=>extent*x);
    segment(end.map(x=>-x),end,'#2467a0',2.5,[3,6]);
    const O=screen([0,0]); ctx.beginPath(); ctx.arc(...O,3,0,2*Math.PI); ctx.fillStyle='#2467a0'; ctx.fill();
    ctx.fillStyle='#555'; ctx.fillText('0',O[0]+8,O[1]+13);
    drawImage(frame.original,'Original');
    drawImage(frame.reflected,'Reflected');
  }
  slider.addEventListener('input',draw);
  picture.onload=draw;
  picture.src='data:image/png;base64,__IMAGE__';
})();
</script>
"""
canvas_html = (canvas_template
    .replace('__ID__', 'reflection-' + uuid.uuid4().hex)
    .replace('__FRAMES__', json.dumps(frames, separators=(',', ':')))
    .replace('__EXTENT__', str(2 * radius + padding))
    .replace('__WIDTH__', str(width))
    .replace('__HEIGHT__', str(height))
    .replace('__IMAGE__', image_base64))
display(HTML(canvas_html))
````
