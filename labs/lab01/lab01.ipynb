## Storing values with variables

---

Usually we don't want an answer to vanish the moment it appears. We want to **store** it under a name and reuse it later. We do that with the `=` sign. Any line of code with `=` is called an assignment statement.

```python
count = 7 + 9
count
```

Read `count = 7+9` as an instruction to Python: evaluate the expression on the right side of the `=`, and store that result under the name `count`.

The value of a variable is not set in stone. Running another assignment with the same name changes which value that name refers to.

```python
count = 12
```

I can even change `count` in a way that uses `count`!

```python
count = count + 1
count
```

Once a value is stored, every later cell can use it. Run the cell below and watch it reuse `count` from above.

```python
count / 15
```

## Notebook memory and restarting the kernel

---

The **kernel** is the engine that runs Python. You can think of it as the notebook's short-term memory: whenever you run a cell with an assignment statement, such as `count = 12`, the kernel remembers that name and value for later cells. It remembers the order in which you **run** cells, even if that order differs from the order on the page.

```python
# We defined this a few cells ago, but the kernel still remembers it.
count
```

If your Colab session ends or you restart the kernel, the notebook forgets those stored values. The code cells remain on the page, but you need to run them again. When you return to a notebook, run its cells from top to bottom so that every name is defined before it is used. For a reproducible notebook:

1. Don't delete a cell that defines a variable used later.
2. Don't use a variable above the cell where it is defined.
3. Don't overwrite built-in names such as `min`, `max`, or `round`, which already have some meaning (as we will see shortly).

If something seems wrong, first save your notebook. Then use Colab's **Runtime** menu to restart the session (the option may be called **Restart session** or **Restart runtime**). This gives the kernel a fresh start. After restarting, run all cells from the top.

## Functions

---

In high school you used functions like $f(x) = x^2$, or $\sin(x)$, or $\max(a, b)$: you plug in one or more **inputs** and get an **output** back.

Python functions follow the same principle. To use, or **call**, a function, you write the function's name, then its inputs — called **arguments** — in parentheses:

$$\underbrace{\texttt{round}}_{\text{name}}(\underbrace{\texttt{3.14159},\ \texttt{2}}_{\text{arguments}})\;\longrightarrow\; \texttt{3.14}$$

Python comes with several functions already built in. Run the next few cells to see a few in action.

```python
round(3.14159)
```

```python
round(3.14159, 2)
```

```python
abs(-4)
```

```python
max(4, 9, 7)
```

```python
type(2 * 3)
```

```python
type(24 / 4)
```
